```python
from serial import Serial
import time

# Open serial port
ser = Serial(
    port='COM3',  # Change this to your COM port. Check from NI MAX software
    baudrate=9600,
    bytesize=8,
    parity='N',
    stopbits=1,
    timeout=1
)
```

```python
# Test commands
commands = [
    "*RST",  # Reset the instrument
    "*IDN?",  # Query instrument identification
    ":SOUR:FUNC VOLT",  # Set source function to voltage
    ":SOUR:VOLT 1",  # Set source voltage to 1V
    ":SENS:FUNC 'CURR'",  # Set measure function to current
    ":SENS:CURR:RANG:AUTO ON",  # Enable auto range for current measurement
    ":OUTP ON",  # Turn output on
    ":READ?",  # Perform measurement and return result
    ":OUTP OFF"  # Turn output off
]

# Send commands and print responses
for command in commands:
    ser.write((command + "\r\n").encode())
    time.sleep(0.1)  # Wait for instrument to process command
    if "?" in command:
        response = ser.readline().decode().strip()
        print(f"Command: {command}")
        print(f"Response: {response}\n")
    else:
        print(f"Command sent: {command}\n")

# Close serial port
ser.close()
```
